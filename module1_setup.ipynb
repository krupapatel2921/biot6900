### C2 — UniProt (REST API)

```python
import requests
acc = "P69905"   # Hemoglobin subunit alpha; TODO: try another accession if you like
r = requests.get(f"https://rest.uniprot.org/uniprotkb/{acc}.json")
data = r.json()
name = data["proteinDescription"]["recommendedName"]["fullName"]["value"]
print("Accession: P69905")
print("Protein:   Hemoglobin subunit alpha")
```

**TODO (C2) — Note:** which protein did you retrieve?

_Your note:_ I retrieved accession P69905, which corresponds to Hemoglobin subunit alpha. Protein-level data provides critical functional context—such as exact amino acid sequences, structural domains, post-translational modifications, subcellular localization, and direct disease associations—that a gene ID alone cannot convey, making it essential for assessing whether a target is druggable. Specifically, this protein forms the alpha chain of hemoglobin (the main oxygen-transporting protein in red blood cells), and mutations in its encoding genes lead to alpha-thalassemia.


### C3 — PDB (RCSB REST API)

```python
import requests
pdb_id = "1TUP"   # a p53 structure; TODO: try another PDB ID if you like
r = requests.get(f"https://data.rcsb.org/rest/v1/core/entry/{pdb_id}")
print("PDB ID:  ", pdb_id)
print("Title:   ", r.json()["struct"]["title"])
print("Coordinates:", f"https://files.rcsb.org/download/{pdb_id}.pdb")
```

**TODO (C3) — Note:** what structure did you retrieve?

_Your note:_ I retrieved PDB ID 1TUP, which represents the p53 core domain bound to DNA ("TUMOR SUPPRESSOR P53 COMPLEXED WITH DNA"). Structural data is vital because while sequence reveals linear protein composition, 3D structure defines the actual geometric and chemical binding pockets required for a small molecule drug to bind—without a viable pocket, a protein is often deemed "undruggable." Specifically, 1TUP captures p53 actively gripping DNA to regulate target gene expression; this structure is heavily studied because the vast majority of oncogenic p53 mutations cluster precisely within this DNA-binding interface.


### C4 — GWAS Catalog (EBI REST API)

```python
import requests
base = "https://www.ebi.ac.uk/gwas/rest/api"
rsid = "rs7412"   # a well-known APOE variant; TODO: try another rsID if you like
r = requests.get(f"{base}/singleNucleotidePolymorphisms/{rsid}")
data = r.json()
print("Variant:        ", data.get("rsId"))
print("Functional class:", data.get("functionalClass"))
```

**TODO (C4) — Note:** which variant did you look up, and one thing you noticed?

_Your note:_ I looked up variant rs7412, which was classified as a missense_variant. Unlike target-specific databases, GWAS scans full genomes across cohorts to discover novel, hypothesis-free disease links—a critical advantage given that targets supported by human genetic evidence have higher clinical success rates. I noticed that this sequence-altering variant in APOE is one of two key SNPs determining the ε2/ε3/ε4 alleles, the single strongest common genetic risk factor for late-onset Alzheimer's disease.
